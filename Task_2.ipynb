```python
# Q1.
def convert_temperature(value, scale):
    if scale.lower() == "c":
        f = (value * 9/5) + 32
        print(f"{value}°C = {f:.2f}°F")
        if value < 10: print("Cold")
        elif value <= 28: print("Mild")
        else: print("Hot")
    elif scale.lower() == "f":
        c = (value - 32) * 5/9
        print(f"{value}°F = {c:.2f}°C")
        if c < 10: print("Cold")
        elif c <= 28: print("Mild")
        else: print("Hot")
    else:
        raise ValueError("Scale must be 'C' or 'F'")

convert_temperature(0, "C")
convert_temperature(86, "F")


# Q2.
def analyze_numbers(numbers):
    total_count = 0
    sum_values = 0
    even_count = 0
    odd_count = 0
    min_val = None
    max_val = None
    for n in numbers:
        total_count += 1
        sum_values += n
        if n % 2 == 0: even_count += 1
        else: odd_count += 1
        if min_val is None or n < min_val: min_val = n
        if max_val is None or n > max_val: max_val = n
    return {
        "total_count": total_count,
        "sum_values": sum_values,
        "even_count": even_count,
        "odd_count": odd_count,
        "min_val": min_val,
        "max_val": max_val
    }

print("Q2:", analyze_numbers([3,5,2,8,10]))


# Q3
def print_diamond(n):
    if n % 2 == 0:
        raise ValueError("n must be odd for a symmetric diamond")
    mid = n // 2
    for i in range(n):
        spaces = abs(mid - i)
        stars = n - 2*spaces
        print(" " * spaces + "*" * stars)

print("Q3:")
print_diamond(5)


# Q4
def find_primes_in_range(start, end):
    primes = []
    for num in range(start, end+1):
        if num < 2: continue
        is_prime = True
        for i in range(2, int(num**0.5)+1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime: primes.append(num)
    return primes

print("Q4:", find_primes_in_range(10, 30))


# Q5
def fibonacci(n):
    if n == 0: return []
    if n == 1: return [0]
    seq = [0,1]
    for i in range(2,n):
        seq.append(seq[-1]+seq[-2])
    return seq

print("Q5:", fibonacci(10))


# Q6
import string
def clean_and_count(text):
    for p in string.punctuation:
        text = text.replace(p,"")
    words = text.lower().split()
    freq = {}
    for w in words:
        freq[w] = freq.get(w,0)+1
    return dict(sorted(freq.items(), key=lambda x:x[1], reverse=True))

print("Q6:", clean_and_count("Hello, hello world! World world."))


# Q7
def flatten_and_unique(nested_list):
    flat = []
    def helper(lst):
        for item in lst:
            if isinstance(item,list):
                helper(item)
            else:
                if item not in flat:
                    flat.append(item)
    helper(nested_list)
    return flat

print("Q7:", flatten_and_unique([1,[2,3,[4,1]],5]))


# Q8
def transpose_matrix(matrix):
    return [[matrix[r][c] for r in range(len(matrix))] for c in range(len(matrix[0]))]

def add_matrices(mat_a, mat_b):
    return [[mat_a[r][c]+mat_b[r][c] for c in range(len(mat_a[0]))] for r in range(len(mat_a))]

print("Q8 Transpose:", transpose_matrix([[1,2,3],[4,5,6]]))
print("Q8 Add:", add_matrices([[1,2],[3,4]], [[5,6],[7,8]]))


# Q9
def aggregate_sales(sales_list):
    summary = {}
    for store in sales_list:
        for item,qty in store.items():
            summary[item] = summary.get(item,0)+qty
    return summary

print("Q9:", aggregate_sales([{"apple":10,"banana":5},{"apple":15,"orange":8}]))


# Q10
def two_sum(nums,target):
    seen={}
    for i,n in enumerate(nums):
        if target-n in seen:
            return [seen[target-n],i]
        seen[n]=i
    return None

print("Q10:", two_sum([2,7,11,15],9))


# Q11. Functional Grade Filter
students=[{"name":"Alice","marks":[80,90,70]},{"name":"Bob","marks":[60,65,70]}]
above_75=list(map(lambda s:s["name"], filter(lambda s: sum(s["marks"])/len(s["marks"])>75, students)))
print("Q11:", above_75)


# Q12
def calculate_stats(*args, **kwargs):
    operation=kwargs.get("operation","sum")
    scale=kwargs.get("scale",1)
    nums=[a*scale for a in args]
    if operation=="sum": return sum(nums)
    elif operation=="mean": return sum(nums)/len(nums)
    else: return None

print("Q12:", calculate_stats(1,2,3,operation="mean",scale=10))


# Q13
def parse_file(filepath):
    valid=[]
    try:
        with open(filepath) as f:
            for line in f:
                try:
                    valid.append(float(line.strip()))
                except ValueError:
                    with open("error.log","a") as log:
                        log.write(f"Invalid line: {line}\n")
    except FileNotFoundError:
        print("File not found")
        return None
    return sum(valid)/len(valid) if valid else None


# Q14
def recursive_search(data,target):
    if isinstance(data,list):
        for item in data:
            if recursive_search(item,target): return True
        return False
    else:
        return data==target

print("Q14:", recursive_search([1,[2,[3,4]]],3))


# Q15
class InsufficientFundsError(Exception): pass
class BankAccount:
    def __init__(self): self._balance=0
    def deposit(self,amt): self._balance+=amt
    def withdraw(self,amt):
        if amt>self._balance: raise InsufficientFundsError("Not enough funds")
        self._balance-=amt
    def get_balance(self): return self._balance
class SavingsAccount(BankAccount):
    def __init__(self,rate): super().__init__(); self.interest_rate=rate
    def apply_interest(self): self._balance+=self._balance*self.interest_rate

acc=SavingsAccount(0.1); acc.deposit(100); acc.apply_interest()
print("Q15:", acc.get_balance())


# Q16
class Item:
    def __init__(self,name,price,qty): self.name,self.price,self.qty=name,price,qty
    def get_total_price(self): return self.price*self.qty
class ShoppingCart:
    def __init__(self): self.items=[]
    def add_item(self,item): self.items.append(item)
    def remove_item(self,name): self.items=[i for i in self.items if i.name!=name]
    def calculate_grand_total(self,tax=0.0):
        subtotal=sum(i.get_total_price() for i in self.items)
        return subtotal*(1+tax)

cart=ShoppingCart(); cart.add_item(Item("Book",100,2)); cart.add_item(Item("Pen",10,5))
print("Q16:", cart.calculate_grand_total(tax=0.1))


# Q18.
class Vector2D:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def __add__(self, other):
        return Vector2D(self.x+other.x, self.y+other.y)
    def __sub__(self, other):
        return Vector2D(self.x-other.x, self.y-other.y)
    def __mul__(self, other):
        if isinstance(other,(int,float)):
            return Vector2D(self.x*other, self.y*other)
        elif isinstance(other,Vector2D):
            return self.x*other.x + self.y*other.y
    def __str__(self):
        return f"Vector({self.x}, {self.y})"
v1, v2 = Vector2D(2,3), Vector2D(4,5)
print("Q18 Add:", v1+v2)
print("Q18 Sub:", v1-v2)
print("Q18 Scalar Mul:", v1*2)
print("Q18 Dot Product:", v1*v2)


# Q19.
class DataPoint:
    def __init__(self, id, feature_vector, label):
        self.id, self.feature_vector, self.label = id, feature_vector, label
    def normalize(self):
        vals = [v for v in self.feature_vector if v is not None]
        xmin, xmax = min(vals), max(vals)
        self.feature_vector = [(x-xmin)/(xmax-xmin) if x is not None else None for x in self.feature_vector]
    def impute_missing(self, fill_value):
        self.feature_vector = [x if x is not None else fill_value for x in self.feature_vector]
dp = DataPoint(1,[10,None,30,50],"A")
dp.impute_missing(20)
dp.normalize()
print("Q19:", dp.feature_vector)


# Q20.
import statistics

class SimpleSeries:
    def __init__(self, data, index=None):
        self.data = data
        self.index = index if index else list(range(len(data)))
    def mean(self):
        return sum(self.data)/len(self.data)
    def std(self):
        return statistics.stdev(self.data)
    def filter_above(self, threshold):
        filtered_data = [d for d in self.data if d>threshold]
        filtered_index = [self.index[i] for i,d in enumerate(self.data) if d>threshold]
        return SimpleSeries(filtered_data, filtered_index)
    def head(self,n):
        return list(zip(self.index[:n], self.data[:n]))
s = SimpleSeries([10,20,30,40], ["a","b","c","d"])
print("Q20 Mean:", s.mean())
print("Q20 Std:", s.std())
print("Q20 Filter > 25:", s.filter_above(25).data)
print("Q20 Head(2):", s.head(2))
```

```python
# Q21.
def compress_string(s):
    result = []
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            result.append(f"{s[i-1]}{count}")
            count = 1
    result.append(f"{s[-1]}{count}")
    compressed = "".join(result)
    return compressed if len(compressed) < len(s) else s
print("Q21:", compress_string("aabcccccaaa"))


# Q22.
def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0
    for n in num_set:
        if n - 1 not in num_set:
            length = 1
            while n + length in num_set:
                length += 1
            longest = max(longest, length)
    return longest
print("Q22:", longest_consecutive([100, 4, 200, 1, 3, 2]))


# Q23.
def moving_average(data, window_size):
    result = []
    window_sum = sum(data[:window_size])
    result.append(round(window_sum / window_size, 2))
    for i in range(window_size, len(data)):
        window_sum += data[i] - data[i - window_size]
        result.append(round(window_sum / window_size, 2))
    return result
print("Q23:", moving_average([1,2,3,4,5,6], 3))


# Q24.
def generate_pascals_triangle(num_rows):
    triangle = [[1]]
    for _ in range(1, num_rows):
        prev = triangle[-1]
        row = [1] + [prev[i] + prev[i+1] for i in range(len(prev)-1)] + [1]
        triangle.append(row)
    return triangle
print("Q24:", generate_pascals_triangle(5))


# Q25.
def group_anagrams(words):
    from collections import defaultdict
    groups = defaultdict(list)
    for word in words:
        key = "".join(sorted(word))
        groups[key].append(word)
    return dict(groups)
print("Q25:", group_anagrams(["eat","tea","tan","ate","nat","bat"]))


# Q26.
def align_records(records, default_val=None):
    all_keys = set().union(*records)
    return [{k: rec.get(k, default_val) for k in all_keys} for rec in records]
print("Q26:", align_records([{"a":1},{"b":2,"a":3}]))


# Q27.
class DimensionMismatchError(Exception): pass
def matrix_multiply(A, B):
    if len(A[0]) != len(B):
        raise DimensionMismatchError("Invalid dimensions")
    return [[sum(a*b for a,b in zip(row,col)) for col in zip(*B)] for row in A]
print("Q27:", matrix_multiply([[1,2],[3,4]], [[5],[6]]))


# Q28.
def extract_nested_keys(d, target_key):
    results = []
    if isinstance(d, dict):
        for k, v in d.items():
            if k == target_key:
                results.append(v)
            results.extend(extract_nested_keys(v, target_key))
    elif isinstance(d, list):
        for item in d:
            results.extend(extract_nested_keys(item, target_key))
    return results
print("Q28:", extract_nested_keys({"a":{"b":1},"b":2}, "b"))


# Q29.
def pivot_summary(data, group_by_col, value_col):
    from collections import defaultdict
    groups = defaultdict(list)
    for row in data:
        groups[row[group_by_col]].append(row[value_col])
    return {g: {"sum":sum(vals),"mean":sum(vals)/len(vals),"count":len(vals)} for g,vals in groups.items()}
print("Q29:", pivot_summary([{"dept":"IT","salary":100},{"dept":"IT","salary":200},{"dept":"HR","salary":150}], "dept","salary"))


# Q30.
import heapq
def process_tasks(tasks):
    heap = [(t["priority"], t["name"]) for t in tasks]
    heapq.heapify(heap)
    return [heapq.heappop(heap)[1] for _ in range(len(heap))]
print("Q30:", process_tasks([{"name":"task1","priority":2},{"name":"task2","priority":1}]))


# Q31.
import time
def time_execution(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} executed in {(end-start)*1000:.2f} ms")
        return result
    return wrapper
@time_execution
def sample_func(n): return sum(range(n))
sample_func(10000)


# Q32.
def chunk_file_reader(filepath, chunk_size=2):
    with open(filepath) as f:
        chunk = []
        for line in f:
            chunk.append(line.strip())
            if len(chunk) == chunk_size:
                yield chunk
                chunk = []
        if chunk: yield chunk


# Q33.
def enforce_types(type_schema):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for arg_name, arg_type in type_schema.items():
                if arg_name in kwargs and not isinstance(kwargs[arg_name], arg_type):
                    raise TypeError(f"{arg_name} must be {arg_type}")
            return func(*args, **kwargs)
        return wrapper
    return decorator
@enforce_types({"val": float})
def test_func(val): return val*2
print("Q33:", test_func(val=2.5))


# Q34.
def memoize(func):
    cache = {}
    def wrapper(*args):
        if args not in cache:
            cache[args] = func(*args)
        return cache[args]
    return wrapper
@memoize
def fib(n): return n if n<2 else fib(n-1)+fib(n-2)
print("Q34:", fib(10))


# Q35.
class Array2D:
    def __init__(self, data, rows, cols):
        self.data, self.rows, self.cols = data, rows, cols
    def __getitem__(self, idx): r,c=idx; return self.data[r*self.cols+c]
    def __setitem__(self, idx, val): r,c=idx; self.data[r*self.cols+c]=val
    def reshape(self, new_rows,new_cols):
        if new_rows*new_cols!=len(self.data): raise ValueError("Invalid reshape")
        return Array2D(self.data,new_rows,new_cols)
arr=Array2D([1,2,3,4],2,2)
print("Q35:", arr[0,1]); arr[1,0]=9; print(arr.data)


# Q36.
class BasePipelineStep:
    def process(self,data): raise NotImplementedError
class FillMissingStep(BasePipelineStep):
    def __init__(self,default_val): self.default_val=default_val
    def process(self,data): return [x if x is not None else self.default_val for x in data]
class ScaleStep(BasePipelineStep):
    def __init__(self,factor): self.factor=factor
    def process(self,data): return [x*self.factor for x in data]
class FilterOutliersStep(BasePipelineStep):
    def __init__(self,min_val,max_val): self.min_val,self.max_val=min_val,max_val
    def process(self,data): return [x for x in data if self.min_val<=x<=self.max_val]
class Pipeline:
    def __init__(self,steps): self.steps=steps
    def execute(self,data):
        for step in self.steps: data=step.process(data)
        return data
pipe=Pipeline([FillMissingStep(0),ScaleStep(2),FilterOutliersStep(0,10)])
print("Q36:", pipe.execute([None,3,20]))


# Q37.
class AccountLedger:
    def __init__(self): self._transactions=[]
    def add_transaction(self,amount): self._transactions.append(amount)
    @property
    def balance(self): return sum(self._transactions)
    @property
    def transaction_count(self): return len(self._transactions)
ledger=AccountLedger(); ledger.add_transaction(100); ledger.add_transaction(-40)
print("Q37:", ledger.balance, ledger.transaction_count)
```

```python
# Q39
class MiniDataFrame:
    def __init__(self, data):
        self.data = data
    def info(self):
        for col, vals in self.data.items():
            non_null = sum(v is not None for v in vals)
            dtype = type(next((v for v in vals if v is not None), None))
            print(f"{col}: {non_null} non-null, dtype={dtype.__name__}")
    def describe(self):
        for col, vals in self.data.items():
            numeric_vals = [v for v in vals if isinstance(v, (int, float))]
            if numeric_vals:
                print(f"{col}: count={len(numeric_vals)}, mean={sum(numeric_vals)/len(numeric_vals):.2f}, "
                      f"min={min(numeric_vals)}, max={max(numeric_vals)}")
    def filter_by(self, col_name, predicate_func):
        filtered = {c: [] for c in self.data}
        for i, val in enumerate(self.data[col_name]):
            if predicate_func(val):
                for c in self.data:
                    filtered[c].append(self.data[c][i])
        return MiniDataFrame(filtered)
df = MiniDataFrame({"age":[25,30,None], "salary":[50000,60000,70000]})
print("Q39 Info:")
df.info()
print("Q39 Describe:")
df.describe()
print("Q39 Filter (age > 25):")
filtered_df = df.filter_by("age", lambda x: x is not None and x > 25)
filtered_df.info()


# Q40
class GroupedData:
    def __init__(self, col_name, data):
        self.col_name = col_name
        self.data = data
        self.groups = {}
        for i, val in enumerate(data[col_name]):
            self.groups.setdefault(val, []).append(i)
    def sum(self, target_col):
        return {g: sum(self.data[target_col][i] for i in idxs) for g, idxs in self.groups.items()}
    def mean(self, target_col):
        return {g: sum(self.data[target_col][i] for i in idxs)/len(idxs) for g, idxs in self.groups.items()}

class MiniDataFrameWithGroup(MiniDataFrame):
    def groupby(self, col_name):
        return GroupedData(col_name, self.data)
df2 = MiniDataFrameWithGroup({"dept":["IT","IT","HR"], "salary":[100,200,150]})
grouped = df2.groupby("dept")
print("Q40 Sum:", grouped.sum("salary"))
print("Q40 Mean:", grouped.mean("salary"))
```
