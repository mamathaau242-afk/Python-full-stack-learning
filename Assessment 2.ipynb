Q1.
Given an integer array `nums`, return all the triplets `[nums[i], nums[j], nums[k]]` such
that `i != j`, `i != k`, and `j != k`, and `nums[i] + nums[j] + nums[k] == 0`. Notice that the solution
set must not contain duplicate triplets.

```python
class Solution:
   def threeSum(self, nums: list[int]) -> list[list[int]]:
       nums.sort()
       result = []
       n = len(nums)
       for i in range(n - 2):
           # Skip duplicate i
           if i > 0 and nums[i] == nums[i-1]:
               continue
           left, right = i + 1, n - 1
           target = -nums[i]
           while left < right:
               current_sum = nums[left] + nums[right]
               if current_sum == target:
                   result.append([nums[i], nums[left], nums[right]])
                   # Skip duplicate left
                   while left < right and nums[left] == nums[left+1]:
                       left += 1
                   # Skip duplicate right
                   while left < right and nums[right] == nums[right-1]:
                       right -= 1
                   left += 1
                   right -= 1
               elif current_sum < target:
                   left += 1
               else:
                   right -= 1
       return result
```

 Q2.
 Implement a function `safe_division` that takes two string arguments, `num1_str` and
`num2_str`, representing numbers. The function should attempt to convert these strings to floats
and then perform division. Handle `ValueError` if the strings cannot be converted to valid
numbers, and `ZeroDivisionError` if the second number is zero. Return the result of the division
as a string, or an appropriate error message if an exception occurs.

```python
def safe_division(num1_str, num2_str):
   try:
       num1 = float(num1_str)
       num2 = float(num2_str)
       result = num1 / num2
       return str(result)
   except ValueError:
       return "Error: Invalid input. Please provide valid numbers."
   except ZeroDivisionError:
       return "Error: Cannot divide by zero."
if __name__ == '__main__':
   num1_str = input()
   num2_str = input()
   print(safe_division(num1_str, num2_str))
```

Q3.
Given two tuples, tuple1 and tuple2, determine if they are 'anagrams' of each other.
This means they contain the same elements with the same frequencies, but possibly in a
different order. The elements within the tuples can be of any hashable type.
Your Answer : from collections import Counter

```python
from collections import Counter
def are_tuple_anagrams(tuple1, tuple2):
   if len(tuple1) != len(tuple2):
       return False
   return Counter(tuple1) == Counter(tuple2)
if __name__ == '__main__':
   len1 = int(input())
   elements1 = []
   for _ in range(len1):
       elements1.append(input())
   t1 = tuple(elements1)
   len2 = int(input())
   elements2 = []
   for _ in range(len2):
       elements2.append(input())
       t2 = tuple(elements2)
   result = are_tuple_anagrams(t1, t2)
   print(str(result).lower())
```

Q4. Given an encoded string, return its decoded string. The encoding rule is:
k[encoded_string], where the encoded_string inside the square brackets is being repeated
exactly k times. k is guaranteed to be a positive integer. You may assume that the input string is
always valid; no extra white spaces, square brackets are well-formed, etc. Furthermore, you may
assume that the original data does not contain any digits and that digits are only for those repeat
numbers, i.e. there are no digits in the original string. The string contains only lowercase English
letters and square brackets.

```python
class Solution:
   def decodeString(self, s: str) -> str:
       stack = [] # Stores [num, current_string] pairs or characters
       current_num = 0
       current_string = ""
       for char in s:
           if char.isdigit():
               current_num = current_num * 10 + int(char)
           elif char == '[':
               stack.append(current_string)
               stack.append(current_num)
               current_string = ""
               current_num = 0
           elif char == ']':
             num = stack.pop()
             prev_string = stack.pop()
             current_string = prev_string + current_string * num
           else:
               current_string += char
       return current_string
```

Q5. Given an integer array `nums`, return the maximum difference between two
successive elements in its sorted form. If the array contains less than two elements, return 0.
You must write an algorithm that runs in linear time.
Your Answer : def maximumGap(nums):

```python
def maximumGap(nums):
   n = len(nums)
   if n < 2:
       return 0
   min_val = min(nums)
   max_val = max(nums)
   if min_val == max_val:
       return 0
bucket_size = max(1, (max_val - min_val) // (n - 1))
num_buckets = (max_val - min_val) // bucket_size + 1
buckets = [[float('inf'), float('-inf')] for _ in range(num_buckets)]
for num in nums:
       if num == min_val or num == max_val:
           continue
       idx = (num - min_val) // bucket_size
       buckets[idx][0] = min(buckets[idx][0], num)
       buckets[idx][1] = max(buckets[idx][1], num)
max_gap = 0
prev_max = min_val
for i in range(num_buckets):
   if buckets[i][0] == float('inf'):
           continue
max_gap = max(max_gap, buckets[i][0] - prev_max)
prev_max = buckets[i][1]
max_gap = max(max_gap, max_val - prev_max)
return max_gap
if __name__ == '__main__':
   nums_str = input()
   nums = [int(x) for x in nums_str.split()]
   result = maximumGap(nums)
   print(result)
```
