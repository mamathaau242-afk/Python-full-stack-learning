 Q1. Implement a Python function `get_value_from_dict` that takes a dictionary and a key.
It should return the value associated with the key. If the key is not found, return the string 'Key
not found'.

```python
def get_value_from_dict(data_dict, key):
   return data_dict.get(key, 'Key not found')
if __name__ == '__main__':
   data_dict_str = input()
   key = input()
   data_dict = eval(data_dict_str)
   result = get_value_from_dict(data_dict, key)
   print(result)
```

Q2. Implement a Python function `is_palindrome(s)` that checks if a given string `s` is a
palindrome (reads the same forwards and backwards, ignoring case and non-alphanumeric
characters). Return `True` if it is, `False` otherwise.

```python
import re
def is_palindrome(s):
   # Convert to lowercase and remove non-alphanumeric characters
   processed_s = re.sub(r'[^a-z0-9]', '', s.lower())
   return processed_s == processed_s[::-1]
```

 Q3. Given a list of integers, count how many elements are strictly greater than their
immediate preceding element. The first element has no preceding element.

```python
 def count_greater_than_preceding(numbers):
   if len(numbers) < 2:
       return 0
   count = 0
   for i in range(1, len(numbers)):
       if numbers[i] > numbers[i-1]:
           count += 1
   return count
if __name__ == '__main__':
   n = int(input())
   numbers_str = input().split()
   numbers = [int(x) for x in numbers_str]
   result = count_greater_than_preceding(numbers)
   print(result)
```

Q4. Write a Python function `filter_even_numbers` that takes a list of integers as input
and returns a new list containing only the even numbers from the original list, maintaining their
relative order.

```python
def filter_even_numbers(numbers):
   even_numbers = []
   for num in numbers:
       if num % 2 == 0:
           even_numbers.append(num)
   return even_numbers
if __name__ == '__main__':
   line1 = input()
   numbers_str = line1.split()
   numbers = []
   if numbers_str and numbers_str[0]: # Check if the split result is not empty or contains an empty
string
       numbers = [int(x) for x in numbers_str]
   result = filter_even_numbers(numbers)
   print(' '.join(map(str, result)))
```

 Q5. Write a function that takes a space-separated string of integers, converts it into a
tuple, and returns 'True' if the tuple is empty, 'False' otherwise.

```python
def is_tuple_empty(s):
   t = tuple(int(x) for x in s.split() if x.strip())
   return str(len(t) == 0)
input_str = input()
result = is_tuple_empty(input_str)
print(result)
```
