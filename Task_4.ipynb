```python
# Part A
import numpy as np

# Task 1
matrix = np.random.randint(1, 51, (5, 5))
print("Matrix:\n", matrix)

# Task 2
print("Numbers > 25:\n", matrix[matrix > 25])

# Task 3
modified_matrix = matrix.copy()
modified_matrix[modified_matrix % 2 == 0] = 0
print("Even numbers replaced:\n", modified_matrix)

# Task 4
arr = np.random.randint(1, 100, 10)
normalized = (arr - arr.min()) / (arr.max() - arr.min())
print("Normalized array:\n", normalized)
```

```python
import pandas as pd
import numpy as np

# Task 1
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Ravi', 'Priya', 'Amit', 'Sneha', 'Kiran', 'Meena'],
    'Department': ['IT', 'HR', 'Finance', 'IT', 'HR', 'Finance'],
    'Age': [25, 30, 28, 35, 40, 29],
    'Salary': [50000, np.nan, 55000, 60000, np.nan, 58000]
}
df = pd.DataFrame(data)

# Task 2
print(df.isnull().sum())

# Task 3
df['Salary'] = df.groupby('Department')['Salary'].transform(lambda x: x.fillna(x.median()))

# Task 4
df['Annual_Bonus'] = df['Salary'] * 0.10
print(df.groupby('Department')['Salary'].mean())
```

```python
data = {
    'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Printer'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories', 'Accessories'],
    'Price': [800, 300, 600, 150, 700],
    'Stock': [10, 100, 30, 15, 5]
}
df = pd.DataFrame(data)

# Task 2
print(df[(df['Price'] > 500) & (df['Stock'] < 50)])

# Task 3
df['Restock_Needed'] = df['Stock'] < 20

# Task 4
print(df.sort_values(by='Price', ascending=False))
```
